```python
# ==========================================
# CELL 1 & 2: Imports & Initialization
# ==========================================
import os
import re
import nltk
import logging
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import spacy
import torch  # Added for BERT

from scipy.stats import chi2_contingency
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import LabelEncoder
from sklearn.utils import resample
from scipy.sparse import hstack
from imblearn.over_sampling import SMOTE
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import stopwords
import string
import joblib

# Replacing VADER imports with Hugging Face Transformers
from transformers import AutoTokenizer, AutoModel

logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Check for GPU availability to speed up BERT extraction
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
logger.info(f"Using device: {device}")
file_path = "D:\\Combined Data.csv"
df = pd.read_csv(file_path)
print(df['status'].unique())
print(df.isnull().sum())

stress_map = {
    'Normal': 'low',
    'Anxiety': 'Medium',
    'Stress': 'Medium',
    'Depression': 'High',
    'Suicidal': 'Extreme',
    'Bipolar': 'Extreme',
    'Personality disorder': 'Extreme',
}

df['stress_level'] = df['status'].map(stress_map)
df = df.dropna(subset=['stress_level'])
print(df['stress_level'].value_counts())


# ==========================================
# CELL 6, 7 & 8: Preprocessing Tasks
# ==========================================
print(df.head())

stemmer = PorterStemmer()
stop_words = set(stopwords.words('english'))

def preprocess_text_tfidf(text):
    text = str(text).lower()
    text = re.sub(r'http\S+|www\S+|https\S+', '', text)
    text = re.sub(r'@\w+|#\w+', '', text)
    text = re.sub(r'\d+', '', text)
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    words = text.split()
    words = [stemmer.stem(w) for w in words if w not in stop_words]
    return " ".join(words)

# Renamed from preprocess_text_vader to bert
def preprocess_text_bert(text):
    text = str(text)
    text = re.sub(r'http\S+|www\S+|https\S+', '', text)
    text = re.sub(r'@\w+|#\w+', '', text)
    return text

df['clean_text'] = df['statement'].apply(preprocess_text_tfidf)
df['bert_text'] = df['statement'].apply(preprocess_text_bert)
df = df[df['clean_text'].str.strip().str.len() > 0]

print("\nSample preprocessed texts:")
print(df[['statement', 'clean_text','bert_text']].head())
print(df['stress_level'].value_counts())


# ==========================================
# CELL 9: BERT Embedding Extraction
# ==========================================
# Initialize BERT tokenizer and model
model_name = "sentence-transformers/all-MiniLM-L6-v2"
tokenizer = AutoTokenizer.from_pretrained(model_name)
bert_model = AutoModel.from_pretrained(model_name).to(device)

def get_bert_embeddings(text_series, batch_size=128):
    """Extracts BERT embeddings in batches using mean pooling."""
    bert_model.eval()
    embeddings = []
    
    for i in range(0, len(text_series), batch_size):
        batch = text_series.iloc[i:i+batch_size].tolist()
        
        encoded_input = tokenizer(batch, padding=True, truncation=True, return_tensors='pt', max_length=128).to(device)
        
        with torch.no_grad():
            model_output = bert_model(**encoded_input)
            
        # Mean Pooling - Take attention mask into account for correct averaging
        attention_mask = encoded_input['attention_mask']
        token_embeddings = model_output[0]
        input_mask_expanded = attention_mask.unsqueeze(-1).expand(token_embeddings.size()).float()
        
        sum_embeddings = torch.sum(token_embeddings * input_mask_expanded, 1)
        sum_mask = torch.clamp(input_mask_expanded.sum(1), min=1e-9)
        batch_embeddings = sum_embeddings / sum_mask
        
        embeddings.extend(batch_embeddings.cpu().numpy())
        
    return np.array(embeddings)

print("\nExtracting BERT embeddings... This may take a few moments.")
bert_features_array = get_bert_embeddings(df['bert_text'])
bert_df = pd.DataFrame(bert_features_array)

print("\nBERT features shape:")
print(bert_df.shape)


# ==========================================
# CELL 10: Train-Test Split
# ==========================================
X_text = df['clean_text'].reset_index(drop=True)
y = df['stress_level'].reset_index(drop=True)
bert_features = bert_df.reset_index(drop=True)

X_train_text, X_test_text, y_train, y_test, bert_train, bert_test = train_test_split(
    X_text, y, bert_features,
    test_size=0.2,
    random_state=42,
    stratify=y
)


# ==========================================
# CELL 11: TF-IDF Vectorization
# ==========================================
vectorizer = TfidfVectorizer(
    max_features=10000,
    ngram_range=(1,3),
    stop_words=None,
    sublinear_tf=True,
    min_df=2
)

X_train_tfidf = vectorizer.fit_transform(X_train_text)
X_test_tfidf = vectorizer.transform(X_test_text)


# ==========================================
# CELL 12: Feature Concatenation (TF-IDF + BERT)
# ==========================================
X_train = hstack([X_train_tfidf, bert_train.values])
X_test = hstack([X_test_tfidf, bert_test.values])
logger.info(f"Combined feature shape - Train:{X_train.shape}, Test:{X_test.shape}")


# ==========================================
# CELL 13: SMOTE Balancing
# ==========================================
try:
    smote = SMOTE(random_state=42, k_neighbors=5)
    X_train_balanced, y_train_balanced = smote.fit_resample(X_train, y_train)
    logger.info(f"After SMOTE - Train distribution:\n{pd.Series(y_train_balanced).value_counts()}")
    print(f"\nAfter SMOTE resampling:")
    print(pd.Series(y_train_balanced).value_counts())
except Exception as e:
    logger.warning(f"SMOTE failed: {e}. Using original training data.")
    X_train_balanced = X_train
    y_train_balanced = y_train


# ==========================================
# CELL 14: Model Training (Random Forest)
# ==========================================
model = RandomForestClassifier(
    n_estimators=300,
    n_jobs=-1,
    max_depth=70,
    min_samples_leaf=2,
    random_state=42,
    class_weight="balanced"
)

model.fit(X_train_balanced, y_train_balanced)
y_pred = model.predict(X_test)
 
accuracy = accuracy_score(y_test, y_pred)
logger.info(f"Test Accuracy: {accuracy:.4f}")


# ==========================================
# CELL 15 & 16: Evaluation Metrics
# ==========================================
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
cm = confusion_matrix(y_test, y_pred, labels=["low", "Medium", "High", "Extreme"])
cm_df = pd.DataFrame(cm,
    index=["Actual: Low", "Actual: Medium", "Actual: High", "Actual: Extreme"],
    columns=["Pred: Low", "Pred: Medium", "Pred: High", "Pred: Extreme"]
)
print(cm_df)


# ==========================================
# CELL 17: Save the Models
# ==========================================
joblib.dump(model, "stress_model.pkl")
joblib.dump(vectorizer, "tfidf_vectorizer.pkl")
# Note: You do not need to save the BERT model via joblib as it is loaded from Hugging Face

print("Model saved successfully!")
```
